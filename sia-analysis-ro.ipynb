# Prepare municipality-level data for Rondônia

This notebook filters the national outpatient dataset to include only records from Rondônia.

The resulting dataset is aggregated by municipality and only used to create the municipality-level choropleth map in the second step of the scrollytelling visualization.

```python
import pandas as pd
```

```python
df = pd.read_csv("sia-data.csv")
df.head()
```

```python
df.loc[df["ano"] == 2025, "mes"].unique()
```

```python
df_ufs = df.groupby("sigla_uf").size().sort_values(ascending=False)
df_ufs
```

```python
df_ro = df[df["sigla_uf"] == "RO"]
df_ro
```

```python
df_ro["nome_municipio"].value_counts()
```

```python
df_ro.to_csv("df_ro_state.csv")
```
