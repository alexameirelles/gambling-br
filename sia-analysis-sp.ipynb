# SIA data analysis

This notebook analyzes the outpatient records extracted from the Outpatient Information System (SIA) to produce the main national indicators used in the project.

The notebook summarizes outpatient visits related to gambling and betting across Brazil between **2015 and 2025**, using the ICD-10 codes:

- **F63.0** — Pathological gambling
- **Z72.6** — Gambling and betting habits

The analysis includes both primary and associated diagnoses.

This notebook produces the datasets and summary statistics used in the story, including:

- National time series of outpatient visits (2015–2025)
- Total number of outpatient visits
- Distribution of ICD-10 codes (primary vs. associated diagnosis)

```python
import pandas as pd
```

```python
df = pd.read_csv("sia-data.csv")
df.head()
```

```python
df.shape
```

```python
df.groupby("ano").size()
```

```python
df.loc[df["ano"] == 2025, "mes"].unique()
```

```python
df.groupby("sexo_paciente").size()
```

```python
df.groupby("raca_cor_paciente").size()
```

```python
df["idade_paciente"].unique()
```

```python
bins = [0, 17, 29, 44, 59, 120]
labels = ["0–17", "18–29", "30–44", "45–59", "60+"]

df["faixa_etaria"] = pd.cut(
    df["idade_paciente"],
    bins=bins,
    labels=labels
)
```

```python
df["faixa_etaria"].value_counts().sort_index()
```

```python
df.groupby("ano").size().sum()
```

```python
df = df.sort_values("ano")
df
```

```python
df.groupby("sigla_uf").size().sort_values(ascending=False)
```

```python
df.value_counts("nome_municipio_cnes")
```

```python
df["id_municipio"].unique()
```

## São Paulo state

Aggregate outpatient visits by municipality to identify where gambling- and betting-related healthcare visits were recorded across the state.

```python
df_sp = df[df["sigla_uf"] == "SP"]
df_sp
```

```python
df_sp["nome_municipio_cnes"].value_counts()
```

```python
df_sp.to_csv("df_sp_state.csv")
```
