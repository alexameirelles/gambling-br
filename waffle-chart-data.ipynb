# Inspect ICD-10 diagnosis fields

This notebook examines how gambling-related ICDs diagnoses are recorded in the outpatient dataset.

The dataset has already been filtered to include records containing gambling-related diagnoses. 

The purpose of this notebook is to verify in which diagnosis fields these codes appear and whether the same record can contain both diagnoses.

```python
import pandas as pd
```

```python
df = pd.read_csv("sia-data.csv")
```

```python
df.head()
```

```python
df.value_counts
```

```python
cids = ["F630", "Z726"]

for cid in cids:
    print(f"\nCID {cid}")
    print("cid_principal_subcategoria:",
          (df["cid_principal_subcategoria"] == cid).sum())
    print("cid_secundario_subcategoria:",
          (df["cid_secundario_subcategoria"] == cid).sum())
    print("cid_causas_associadas_subcategoria:",
          (df["cid_causas_associadas_subcategoria"] == cid).sum())
```

```python
cols = [
    "cid_principal_subcategoria",
    "cid_secundario_subcategoria",
    "cid_causas_associadas_subcategoria",
]

tem_f630 = (
    (df["cid_principal_subcategoria"] == "F630") |
    (df["cid_secundario_subcategoria"] == "F630") |
    (df["cid_causas_associadas_subcategoria"] == "F630")
)

tem_z726 = (
    (df["cid_principal_subcategoria"] == "Z726") |
    (df["cid_secundario_subcategoria"] == "Z726") |
    (df["cid_causas_associadas_subcategoria"] == "Z726")
)

df_ambos = df[tem_f630 & tem_z726]
df_ambos
```
